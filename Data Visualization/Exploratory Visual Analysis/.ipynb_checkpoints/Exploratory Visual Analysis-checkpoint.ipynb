# Visual Exploratory Analysis

In this visual exploratory analysis I will be looking at employment, population, and housing data from 2001 to 2024 in Colorado.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import matplotlib as mpl
from matplotlib import gridspec
```

```python
df = pd.read_csv('Project 1/County Jobs by Sector Lookup.csv')

# cleaning
df['Total Jobs'] = df['Total Jobs'].str.replace(',', '').astype(int)  # convert total jobs from str to int

#replace wordy sectors
df['Job Sector Name'] = df['Job Sector Name'].str.replace('Mining, Quarrying, and Oil and Gas Extraction', 'Mining')
df['Job Sector Name'] = df['Job Sector Name'].str.replace('Administrative and Support and Waste Management and Remediation Services', 'Admin Support and Waste Management')

# drop public admin which has 0 total jobs
df = df[df['Job Sector Name']!='Public Administration']

# seperate totals from sectors
all_sectors = df[df['Job Sector Name'] != 'Total All Sectors']
total_sectors = df[df['Job Sector Name'] == 'Total All Sectors']

# seperate state data from county data  
state_tot_sec = total_sectors[total_sectors['County Name'] == 'Colorado']
county_tot_sec = total_sectors[total_sectors['County Name'] != 'Colorado']
state_all_sec = all_sectors[all_sectors['County Name'] == 'Colorado']
county_all_sec = all_sectors[all_sectors['County Name'] != 'Colorado']

# add total jobs across sector and employment rate of each sector
total_jobs = state_tot_sec[['Year', 'Total Jobs']]
state_sector_rate = state_all_sec.merge(total_jobs, on='Year')
state_sector_rate['Sector Employment Rate'] = state_sector_rate['Total Jobs_x']/state_sector_rate['Total Jobs_y']
```

```python
# alter sector names to fit as titles 
sectors = state_all_sec['Job Sector Name'].unique()
titles = []

for i in sectors:
    if len(i) > 30:
        l = i.split()
        top = " ".join(l[:len(l)//2])
        bottom = " ".join(l[len(l)//2:])
        titles.append(top + '\n' + bottom)
        
    else: titles.append(i)
```

I first want to explore the County Jobs by Sector data. There are three things I am intrested at after looking at the data: job growth of different industries, percentage of employment change in sector, and change in employment across sector. Below I plot the Job Growth of 23 industries from 2001 to 2024. Some industries remane unchanged over the years like federal government, utilities, agriculture/forestry, and information. I found it really intresting that finance and insurance and wholesale and resale trade were also among those unchanging fields. Given growth in consumerism but perhaps those trends stablaized before 2001. Healthcare and Scientific/Technical Services were far and away the fastest and most constant growing industries. I would perhaps also want to look more into why local and state government are growing similarly despite federal government is unchanging. Construction and Food Service expectedly so have the most instability in growth. Construction dips right after the 2008 recession and food service drops significantly around 2020 when social distancing began.

```python
# plot job growth of different industries
fig, axs = plt.subplots(nrows=5, ncols=5, figsize=(10, 10), num=1, sharey=True, sharex=True)

axs = axs.ravel()

for i, ax in enumerate(axs):
    if i > 22: pass
    else: 
        x = state_sector_rate[state_sector_rate['Job Sector Name'] == sectors[i]]
        ax.plot(x['Year'], x['Total Jobs_x'])
        ax.set_title(titles[i], fontsize=8)

plt.suptitle('Job Growth of Different Industries (2001-2024)')
plt.tight_layout()
```

This heat plot shows the change is percentage of employment from 2001 to 2024 sorted descending by percentage in 2024. This plot reveals similar trends to the above plot, but not completly the same. As I would have expected sectors that showed little growth or decline, saw a decrease in rate as other fields grew in count and percentage. Resale trade and federal government sectors show this best. Resale almost looks like its increasing, very slowly albeit but increasing, above. However, there is a clear and quick change in color from pink to purple after ~2021, signaling a drop in employment percentage for retail trade. The federal government sector also shows a pretty steady drop in percentage. 

I spent a lot of time playing with color palettes here but ultimately thought that the hls palette with black at 0 best showed the small changes in the large combination of features. I found the heat plot better than something similar to what I did with total counts but the changes in some. For example the rate vs time plot for the federal government sector line plot was almost impossible to see the change versus on the heat map the change is clear. To see just how much it has changed you have to use the legend on the side, but the legend provides an additional length factor for the eye to use to distinguish. 

```python
# add total jobs across sector and calculte percentage of employment for each sector that year
total_jobs = state_tot_sec[['Year', 'Total Jobs']]
state_sector_rate = state_all_sec.merge(total_jobs, on='Year')
state_sector_rate['Sector Employment Rate'] = state_sector_rate['Total Jobs_x']/state_sector_rate['Total Jobs_y']

# pivot table to have sector name as index and year as columns 
employ_rate = state_sector_rate.pivot(index='Job Sector Name', columns='Year', values='Sector Employment Rate')
employ_rate = employ_rate.fillna(0).sort_values(2024, ascending=False)

# plot heat map of employment rate
fig = plt.figure(figsize=(12, 9))

my_cmap = sns.blend_palette(['#000000','#db5f57', '#db9057', '#dbc257', '#c3db57', '#91db57', '#5fdb57', '#57db80', '#57dbb2', 
                             '#57d3db', '#57a2db', '#5770db', '#6f57db', '#a157db', '#d357db', '#db57b2', '#db5780'], as_cmap=True)

heat = sns.heatmap(employ_rate, cmap=my_cmap)
heat.set_xticks(heat.get_xticks(), heat.get_xticklabels(), rotation=60)
heat.set_title('Change in Sector Employment Rate')
plt.show()
```

```python
# break counties down to sizes
c = county_tot_sec[county_tot_sec['Year'] == 2024].copy()
bins = [0, 1800, 7000, 13000, 49000, float('inf')]
labels = ['XS', 'S', 'M', 'L', 'XL']

# map to all years and add to new data frame
c['County Size'] = pd.cut(c['Total Jobs'], bins=bins, labels=labels)   # fix 
county_size = c[['County Name', 'County Size']]
county_total_sizes = county_tot_sec.merge(county_size, on='County Name')
```

The last plots I want to look at before introducing more data is how job growth changes across counties. Most counties see a fairly steady job growth. A good chunck haven't changed much and very few have decreased in job counts. One thing I would be very intrested in researching is the random spikes in job growth in some counties before dropping and growing like normal again. This primarily includes several mountain town counties like Garfield, Eagle, and Gunnison around 2008-2009 and the begining of growth in many small and xtra small counties around 2013-2014. 

```python
XL = county_total_sizes[county_total_sizes['County Size'] == 'XL']
county_names = XL['County Name'].unique()

fig, axs = plt.subplots(nrows=3, ncols=4, figsize=(10, 8), num=1, sharey=True, sharex=True)

axs = axs.ravel()

for i, ax in enumerate(axs):
    x = XL[XL['County Name'] == county_names[i]]
    ax.bar(x['Year'], x['Total Jobs'])
    ax.set_title(county_names[i], fontsize=8)

plt.suptitle('Total Job Growth of Xtra Counties in Colorado (2001-2024)')
plt.tight_layout()
```

```python
L = county_total_sizes[county_total_sizes['County Size'] == 'L']
county_names = L['County Name'].unique()

fig, axs = plt.subplots(nrows=3, ncols=3, figsize=(8, 8), num=1, sharey=True, sharex=True)

axs = axs.ravel()

for i, ax in enumerate(axs):
    x = L[L['County Name'] == county_names[i]]
    ax.bar(x['Year'], x['Total Jobs'])
    ax.set_title(county_names[i], fontsize=8)

plt.suptitle('Total Job Growth of Large Counties in Colorado (2001-2024)')
plt.tight_layout()
```

```python
M = county_total_sizes[county_total_sizes['County Size'] == 'M']
county_names = M['County Name'].unique()

fig, axs = plt.subplots(nrows=3, ncols=3, figsize=(8, 8), num=1, sharey=True, sharex=True)

axs = axs.ravel()

for i, ax in enumerate(axs):
    x = M[M['County Name'] == county_names[i]]
    ax.bar(x['Year'], x['Total Jobs'])
    ax.set_title(county_names[i], fontsize=8)

plt.suptitle('Total Job Growth of Medium Counties in Colorado (2001-2024)')
plt.tight_layout()
```

This data looks at County Population and Housing Data which I am intrested in looking at how it relates to job growth. In an ideal world job growth would roughly mimick and stay not far below population (100% population - % to young - % to old = total jobs).

```python
df2 = pd.read_csv('Project 1/County Population and Housing Data Lookup.csv')

#selected features to focus on
df2 = df2[['County Name', 'Year', 'Population', 'Household Population', 'Households', 'Household Size', 'Total Housing Units']]

# fix non int values
for i in ['Population', 'Household Population', 'Households', 'Total Housing Units']:
    df2[i] = df2[i].str.replace(',', '').astype(int)

# merge with data from df2
county_yearly_data = county_total_sizes.merge(df2, on=['County Name', 'Year'])

df2.head()
```

```python
# merge with data from df2
county_yearly_data = county_total_sizes.merge(df2, on=['County Name', 'Year'])
county_yearly_data = county_yearly_data.drop(['County FIPS', 'Job Sector Code', 'Job Sector Name'], axis=1)

# create dataframe for colorado totals
tot_job_pop = county_yearly_data.groupby('Year').agg({'Total Jobs':'sum', 'Population':'sum'})
```

First lets look at total job growth across county and sector in combination with population growth over this time period. Population growth and job growth seem to grow at roughly the same rate, job growth might be a little slower overall. However, it seems the total jobs at any given time is about 50% of the total population. I would expect a much smaller percentage of the population to be to old or young to work.

```python
fig, ax = plt.subplots(figsize=(8,6))

sns.lineplot(data=tot_job_pop, x='Year', y='Total Jobs', label='Total Jobs', ax=ax)
sns.lineplot(data=tot_job_pop, x='Year', y='Population', label='Population', ax=ax)

ax.set_ylabel('')

ax.axvline(x=2008, color='red', linewidth=1)
ax.axvline(x=2020, color='red', linewidth=1)
ax.yaxis.get_major_formatter().set_scientific(False)
ax.set_ylim(bottom=0)
ax.legend()
ax.annotate('2008 Financial\nCrisis', xy=(2008, 2690103), xytext=(2010, 3500000), xycoords='data', color='red',
            arrowprops={'facecolor': 'black', 'width': 1, 'headwidth': 4})
ax.annotate('Start of\nCovid', xy=(2020, 3109248), xytext=(2016, 4000000), xycoords='data', color='red', 
           arrowprops={'facecolor': 'black', 'width': 1, 'headwidth': 4})
ax.set_title('Total Job and Population Growth(2001-2024)')
```

Now lets look at total jobs related to population. I broke it down by county size for scale. The red line represents the origin or where total jobs would perfectly mimick population (unexpected). Population is my y-axis so being above the line means population is larger than total jobs (expected).

Note there is 64 counties in colorado and fitting that on a readable legend was not gonna happen. The actual county doesn't really matter here. In further analysis I would break down to Denver metro area, Front Range, and ski town counties.

```python
fig = plt.figure(figsize=(12, 9))

gs = fig.add_gridspec(4, 9)

ax0 = plt.subplot(gs[:2, :3])
ax1 = plt.subplot(gs[:2, 3:6])
ax2 = plt.subplot(gs[:2, 6:])
ax3 = plt.subplot(gs[2:, 1:4])
ax4 = plt.subplot(gs[2:, 5:8])

a = [ax0, ax1, ax2, ax3, ax4]
sizes = ['XS', 'S', 'M', 'L', 'XL']

for i in range(5):
    data = county_yearly_data[county_yearly_data['County Size'] == sizes[i]]
    sns.scatterplot(data=data, x='Total Jobs', y='Population', hue='County Name', ax=a[i])
    a[i].get_legend().remove()
    a[i].tick_params(axis='x', labelrotation=45)
    a[i].plot(data['Total Jobs'], data['Total Jobs'], color='red')
    a[i].set_title(sizes[i] + ' Counties')
    

plt.suptitle('Job Growth vs Household Growth')    
plt.tight_layout()
```
